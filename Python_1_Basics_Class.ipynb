# 1. Introduction to Python

## 1.1 Why Python?

Python is one of the most widely used programming languages in the world. Three reasons
explain most of that popularity:

**1. Simplicity** — Python reads almost like English. There are fewer symbols and less
"ceremony" than most languages, so you spend time solving your problem instead of fighting
syntax.

**2. Readability** — Python *forces* clean layout through indentation. Code written by one
person is usually easy for the next person to read. This matters more than beginners expect:
you read code far more often than you write it.

**3. Ecosystem** — There is a high‑quality, free library for almost anything:
- **Data & analytics:** `pandas`, `numpy`, `matplotlib`
- **Machine learning / AI:** `scikit-learn`, `pytorch`, `transformers`
- **Web:** `django`, `flask`, `fastapi`
- **Automation & scripting:** `requests`, `openpyxl`, `selenium`

You rarely start from scratch — you assemble existing building blocks.

## 1.3 Two ways to run Python: script vs interactive shell

There are two fundamentally different ways to run Python, and knowing when to use each is
important.

### A) The interactive shell (REPL)
**REPL** = Read–Eval–Print Loop. You type one line, Python runs it immediately and shows the
result, then waits for the next line. Start it by typing `python` in a terminal:

```text
>>> 2 + 3
5
>>> name = "Sam"
>>> name.upper()
'SAM'
>>> exit()
```

✅ Great for: quick experiments, testing an idea, checking "what does this function do?"
❌ Bad for: anything you want to keep — close the window and it's all gone.

### B) Running a script (a `.py` file)
You write your code in a file, e.g. `hello.py`, then run the whole file at once:

```bash
python hello.py
```

✅ Great for: real programs, anything reusable, anything more than a few lines.
❌ You don't see intermediate values automatically — you must `print()` what you want to see.

A **Jupyter notebook** (this file) is a hybrid: like a REPL, each *cell* runs independently and
shows output inline — but like a script, your work is saved. That's exactly why notebooks are
the standard tool for *learning* and *data exploration*.

# 2. Basic Syntax & Data Types

## 2.1 Variables — names for values

A **variable** is a label that points to a value. You create one with `=` (the *assignment*
operator). No need to declare a type first — Python figures it out.

````python
age = 30
age = 30.5
city = "Pune"
price = 19.99

print(type(age))
print(city)
print(price)
````

Things to notice:
- `=` means **"store the value on the right into the name on the left"**. It is *not* the
  mathematical "equals". (We'll meet `==` for comparison in Module 3.)
- A variable can be **reassigned** at any time, even to a different type.

````python
x = 5
print(x)

x = "now I'm text"   # totally allowed in Python
print(x)
````

## 2.2 Constants & naming conventions

Python has **no built‑in "constant"** (a value that can't change). The *convention* is to
write names you intend never to change in **ALL_CAPS**. It's a signal to humans, not enforced
by Python.

````python
PI = 3.14159          # convention: ALL_CAPS = "please don't change me"
MAX_USERS = 100

radius = 4
area = PI * radius * radius
print("Area of circle:", area)
````

### Naming rules (enforced by Python)
A valid variable name:
- may contain **letters, digits, and underscores** (`_`)
- **must not start with a digit**
- **cannot be a reserved keyword** (`if`, `for`, `class`, `True`, …)
- is **case‑sensitive**: `total`, `Total`, and `TOTAL` are three different variables

### Naming style (PEP 8 — the community style guide)
- Variables & functions → `snake_case` → `first_name`, `total_price`
- Constants → `UPPER_SNAKE_CASE` → `MAX_RETRIES`
- Choose **descriptive** names: `n` is mysterious; `number_of_students` explains itself.

````python
# ✅ Good names
first_name = "Asha"
total_price = 249.50
is_logged_in = True

# ⚠️ Legal but unclear — avoid in real code
a = "Asha"
tp = 249.50

# ❌ These would cause ERRORS — uncomment ONE at a time to see the message:
# 2nd_place = "silver"     # can't start with a digit
# class = "Physics"        # 'class' is a reserved keyword
# my-name = "Asha"         # hyphens aren't allowed (Python reads it as my MINUS name)

print(first_name, total_price, is_logged_in)
````

## 2.5 Type casting — converting between types

You can deliberately convert a value from one type to another using `int()`, `float()`,
`str()`, and `bool()`. This is called **casting**.

This matters a lot in real programs because **user input always arrives as a string** — even
when the user typed a number (we'll hit this in Module 4).

````python
# string -> int
text_number = "25"
real_number = int(text_number)
print(real_number, "->", type(real_number))
````

````python
max_marks = 500
student_marks = "200"

percentage = int(student_marks) / max_marks
print(percentage)
print(f"Percentage: {percentage*100}%") # f string is covered later in the notebook
print(type(student_marks))
````

````python
# int -> float, and float -> int (note: converting to int TRUNCATES, it does not round!)
print(float(7))        # 7.0
print(int(7.9))        # 7  (the .9 is chopped off, NOT rounded to 8)
print(round(7.9))      # 8  (use round() if you actually want rounding)
````

````python
# '+' operator: does addition when used with numbers / does concatenation when used with strings

print(4 + 5)
print("Hello" + " " + "World")
````

````python
# number -> string (needed when you want to glue a number onto text with +)
age = 30
message = "I am " + str(age) + " years old."
print(message)
````

# 3. Operators & Expressions

An **operator** is a symbol that performs an operation (like `+` or `>`). An **expression** is
any combination of values and operators that Python can evaluate down to a single value, e.g.
`3 + 4 * 2`.

## 3.1 Arithmetic operators

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | division (always float) | `7 / 2` | `3.5` |
| `//` | floor division (drops remainder) | `7 // 2` | `3` |
| `%` | modulo (the remainder) | `7 % 2` | `1` |
| `**` | exponent (power) | `7 ** 2` | `49` |

````python
a = 7
b = 2

print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)     # 3.5  -> note the decimal: / is always a float
print("a // b =", a // b)    # 3    -> floor division, throws away the fraction
print("a % b  =", a % b)     # 1    -> remainder of 7 divided by 2
print("a ** b =", a ** b)    # 49   -> 7 to the power of 2
````

### Operator precedence (order of operations)
Python follows normal maths rules — **PEMDAS/BODMAS**: Parentheses first, then Exponents, then
Multiply/Divide, then Add/Subtract. When in doubt, **add parentheses** to make intent obvious.

````python
print(3 + 4 * 2)      # 11  -> * happens before +
print((3 + 4) * 2)    # 14  -> parentheses force the addition first
print(2 ** 3 + 1)     # 9   -> exponent first (8), then +1
````

### A practical use of `%` (modulo): even or odd?
A number is **even** if dividing by 2 leaves remainder 0.

````python
number = 18
print("Remainder when divided by 2:", number % 2)
print("Is it even?", number % 2 == 0)
````

## 3.2 Comparison operators

These compare two values and always return a **boolean** (`True`/`False`).

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

> ⚠️ **`=` vs `==`** is the #1 beginner trap. `=` *assigns* a value; `==` *compares* two values.

````python
x = 10
y = 3

print("x == y :", x == y)    # False
print("x != y :", x != y)    # True
print("x >  y :", x > y)     # True
print("x <  y :", x < y)     # False
print("x >= 10:", x >= 10)   # True
print("y <= 3 :", y <= 3)    # True
````

````python
# Comparisons work on strings too (alphabetical / dictionary order):
print("apple" == "apple")    # True
print("apple" < "banana")    # True  -> 'a' comes before 'b'
print("Zoo" < "apple")       # True  -> uppercase letters sort before lowercase!
````

## 3.3 Logical operators

Combine boolean values with `and`, `or`, `not`.

| Operator | True when… |
|----------|-----------|
| `and` | **both** sides are True |
| `or`  | **at least one** side is True |
| `not` | flips True↔False |

````python
age = 25
has_ticket = True

# and -> BOTH must be True
print(age >= 18 and has_ticket)      # True (18+ AND has a ticket)

# or -> at least one True
is_weekend = False
is_holiday = True
print(is_weekend or is_holiday)      # True (it's a holiday)

# not -> flips it
print(not has_ticket)                # False
````

````python
# Real-world style example: can this user enter the event?
age = 16
has_guardian = True

can_enter = (age >= 18) or (age >= 13 and has_guardian)
print("Allowed to enter:", can_enter)   # True: under 18 but 13+ with a guardian
````

## 3.4 Assignment operators

`=` stores a value. The **compound** assignment operators update a variable *based on its own
current value* — a handy shorthand.

| Operator | Example | Equivalent to |
|----------|---------|---------------|
| `=`  | `x = 5`  | — |
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 3` | `x = x - 3` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 3` | `x = x / 3` |
| `//=`| `x //= 3`| `x = x // 3` |
| `%=` | `x %= 3` | `x = x % 3` |
| `**=`| `x **= 3`| `x = x ** 3` |

````python
score = 10
print("start:", score)

score += 5       # add 5    -> 15
print("after += 5:", score)

score -= 3       # subtract -> 12
print("after -= 3:", score)

score *= 2       # double   -> 24
print("after *= 2:", score)

score //= 5      # floor /  -> 4
print("after //= 5:", score)
````

## 3.5 String operations

Strings support their own set of operations and a large family of built‑in **methods**
(functions attached to the value, called with a dot: `value.method()`).

````python
first = "Vineet"
last = "Sharma"

# Concatenation with +  (glue strings together)
full = first + " " + last
print(full)

# Repetition with *
print("=" * 30)         # a quick divider line of 30 equals signs
print("ha" * 3)         # 'hahaha'

# Length with len()
print("Length of full name:", len(full))
````

````python
text = "  Python is Fun  "

print(text.upper())        # ALL CAPS
print(text.lower())        # all lowercase
print(text.strip())        # remove leading/trailing spaces
print(text.replace("Fun", "powerful"))
print("Python" in text)    # True -> substring membership check

print("a" not in "abc")
````

````python
# Indexing & slicing: strings are sequences of characters, counted from 0
word = "PYTHON"
#       012345
print(word[0])      # 'P'  -> first character
print(word[-1])     # 'N'  -> last character (negative counts from the end)
print(word[0:3])    # 'PYT' -> characters 0,1,2 (the end index 3 is excluded)
print(word[2:])     # 'THON' -> from index 2 to the end
````

## 3.6 f‑strings — the modern way to build text

An **f‑string** (formatted string) lets you drop variables and even expressions directly inside
a string by prefixing it with `f` and wrapping the value in `{ }`. This is cleaner than gluing
with `+` and avoids manual `str()` casting.

````python
name = "Asha"
age = 29

# The OLD way (works, but clunky and needs str() casts):
print("Hi " + name + ", you are " + str(age) + " years old.")

# The f-string way (preferred):
print(f"Hi {name}, you are {age} years old.")
````

````python
# You can put any EXPRESSION inside the braces:
price = 250
quantity = 3
print(f"Total cost: {price * quantity}")

# Number formatting: :.2f  -> show exactly 2 decimal places
pi = 3.14159265
print(f"Pi rounded: {pi:.2f}")

# Handy for aligned output / currency:
amount = 1999.5
print(f"Amount due: ₹{amount:,.2f}")    # thousands separator + 2 decimals
````

## 4.3 Activity B — a small calculator

**Goal:** ask for two numbers and an operator (`+ - * /`), then show the result.

We'll build it up in stages so the logic is clear. This introduces `if / elif / else` — a
**decision** structure that runs different code depending on a condition. (You'll cover this in
depth later; here we just use it to pick the right operation.)

````python
# --- Calculator: interactive version ---
print("=== Simple Calculator ===")

a = float(input("Enter first number:  "))
b = float(input("Enter second number: "))
op = input("Choose an operator (+, -, *, /): ")

if op == "+":
    result = a + b
elif op == "-":
    result = a - b
elif op == "*":
    result = a * b
elif op == "/":
    # Guard against the classic division-by-zero crash:
    if b == 0:
        result = "Error: cannot divide by zero"
    else:
        result = a / b
else:
    result = f"Unknown operator: {op}"

print(f"{a} {op} {b} = {result}")
````

````python
# --- Calculator: non-interactive version (edit the three values to test) ---
a = 20.0
b = 4.0
op = "*"

if op == "+":
    result = a + b
elif op == "-":
    result = a - b
elif op == "*":
    result = a * b
elif op == "/":
    result = "Error: cannot divide by zero" if b == 0 else a / b
else:
    result = f"Unknown operator: {op}"

print(f"{a} {op} {b} = {result}")
````

### Walkthrough of what the calculator does
1. **Collect inputs** — two numbers (cast to `float`) and one operator (kept as a string).
2. **Decide** — `if/elif/else` checks which operator the user chose and runs the matching maths.
3. **Handle the edge case** — dividing by zero would crash the program, so we check `b == 0`
   first and return a friendly message instead.
4. **Report** — an f‑string prints the full expression and its answer in one clean line.
